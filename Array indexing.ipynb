NUMPY INDEXING AND SLICING OF ARRAYS

```python
arr= np.arange(11,21)
arr
```

```python
arr[6]
```

```python
arr[1:5]
```

NUMPY INDEXING AND SLICING OF MATRIX

```python
arr1=np.arange(1,31).reshape(6,5)
arr1
```

```python
arr1[5,4] ## first rows and then columns ##
```

```python
arr1[0:2, 1:3] ## range of rows, range of columns  ##
```

```python
arr1[3:6,3:5]
```

```python
arr1[:,2]
```

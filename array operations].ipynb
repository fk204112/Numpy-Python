ADVANCE ARRAY MANIPULATIONS 

STACKING ARRAY

```python
a=np.array([1,2,3,4])
b=np.array([5,6,7,8])
```

```python
np.vstack((a,b)) ## vertical stack ##
```

```python
np.hstack((a,b)) ## horizontal stack ##
```

```python
np.column_stack((a,b))
```

SPLITTING  ARRAYS

```python
c=np.arange(16).reshape(4,4)
c
```

```python
d=np.hsplit(c,4) ## horizontal split of array ##
d
```

```python
e=np.vsplit(c,4) ## vertical split of array ##
e
```

```python
for i in d:
    print(i)
```

```python
for i in e:
    print(i)
```
